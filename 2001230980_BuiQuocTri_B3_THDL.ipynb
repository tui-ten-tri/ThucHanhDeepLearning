```python
# LAB 3 - ARTIFICIAL NEURAL NETWORK (ANN) CHUẨN HÓA DỮ LIỆU & CHỐNG DATA LEAKAGE
import tensorflow as tf
from tensorflow.keras import datasets
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
import numpy as np

# ==========================================
# 1. TẢI VÀ TÁCH DỮ LIỆU ĐẦU TIÊN (SPLIT FIRST)
# ==========================================
# Tuân thủ Golden Rule: Phải có tập dữ liệu độc lập ngay từ đầu
(X_train_raw, y_train), (X_test_raw, y_test) = datasets.mnist.load_data()

print("Kích thước tập Train gốc:", X_train_raw.shape) # (60000, 28, 28)
print("Kích thước tập Test gốc:", X_test_raw.shape)   # (10000, 28, 28)

# ==========================================
# 2. TIỀN XỬ LÝ CHUẨN HÓA (ĐÚNG PIPELINE - TRÁNH LEAKAGE)
# ==========================================
# Mô phỏng quá trình Min-Max Scaling [0, 255] về [0, 1] đúng chuẩn:
# Chỉ tìm Min, Max trên tập Train (Mặc dù ảnh mặc định là 0 và 255, nhưng viết tường minh để hiểu bản chất)
min_val = float(np.min(X_train_raw)) # Học từ tập Train = 0.0
max_val = float(np.max(X_train_raw)) # Học từ tập Train = 255.0

# Áp dụng hàm Transform (chuyển đổi) dựa trên thông số thu được từ tập Train
X_train = (X_train_raw.astype('float32') - min_val) / (max_val - min_val)
X_test = (X_test_raw.astype('float32') - min_val) / (max_val - min_val)

# KHÔNG ĐƯỢC làm: min_test = np.min(X_test_raw) rồi chia riêng cho tập Test!

# ==========================================
# 3. XÂY DỰNG MẠNG NƠ-RON NHÂN TẠO (ANN ARCHITECTURE)
# ==========================================
model = Sequential()

# Tầng 1: Duỗi phẳng (Flatten) ma trận ảnh 2D (28x28) thành Vector 1D (784 phần tử)
model.add(Flatten(input_shape=(28, 28)))

# Tầng 2 (Hidden Layer): 256 nơ-ron, khởi tạo trọng số uniform, hàm kích hoạt relu
model.add(Dense(units=256, kernel_initializer='uniform', activation='relu'))

# Tầng 3 (Output Layer): 10 nơ-ron tương ứng 10 chữ số phân loại, kích hoạt softmax
model.add(Dense(units=10, kernel_initializer='uniform', activation='softmax'))

# Hiển thị tóm tắt cấu trúc mạng ANN đã thiết lập
print("\n--- Cấu trúc mô hình mạng ANN ---")
model.summary()

# ==========================================
# 4. BIÊN DỊCH VÀ HUẤN LUYỆN MÔ HÌNH (TRAINING)
# ==========================================
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print("\n--- Bắt đầu huấn luyện mô hình ANN ---")
# Huấn luyện trong 5 Epochs để chạy demo nghiệm thu nhanh chóng
model.fit(X_train, y_train, epochs=5, batch_size=64, verbose=1)

# ==========================================
# 5. ĐÁNH GIÁ TRÊN TẬP TEST ĐỘC LẬP (CHỈ SỬ DỤNG MỘT LẦN Ở CUỐI)
# ==========================================
print("\n--- Đánh giá mô hình trên tập Test (Sacred Test Set) ---")
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=1)
print(f"Độ chính xác trên tập Test: {test_acc * 100:.2f}%")

# Kiểm tra dự báo mẫu phần tử đầu tiên (Mục 3 trong tài liệu Lab 3)
predictions = model.predict(X_test)
predicted_label = np.argmax(predictions[0])
true_label = y_test[0]

print(f"\nKết quả kiểm tra phần tử đầu tiên của tập Test:")
print(f"-> Nhãn dự đoán (Predicted): {predicted_label}")
print(f"-> Nhãn thực tế (True label): {true_label}")
```
