```python
# TÀI LIỆU LAB 2 - CƠ BẢN TENSORFLOW

import tensorflow as tf

# ==========================================
# 1. KHỞI TẠO TENSOR & ĐỊNH DẠNG (RANK)
# ==========================================
# Rank 0 (Scalar)
rank_0_tensor = tf.constant(4)
print("Rank 0:", rank_0_tensor)

# Rank 1 (Vector)
rank_1_tensor = tf.constant([2.0, 3.0, 4.0])
print("Rank 1:", rank_1_tensor)

# Rank 2 (Matrix)
rank_2_tensor = tf.constant([[1, 2], [3, 4], [5, 6]], dtype=tf.float32)
print("Rank 2:\n", rank_2_tensor)

# Xem thuộc tính Tensor
print("Kích thước (Shape):", rank_2_tensor.shape)
print("Kiểu dữ liệu (Dtype):", rank_2_tensor.dtype)

# ==========================================
# 2. CÁC HÀM TẠO TENSOR TỰ ĐỘNG
# ==========================================
tensor_zeros = tf.zeros([3, 3])
tensor_ones = tf.ones([2, 3])
tensor_fill = tf.fill([2, 2], 9)
tensor_linspace = tf.linspace(5.0, 20.0, 5)
tensor_range = tf.range(3.0, 7.0, delta=0.5)

# Sinh số ngẫu nhiên
tensor_normal = tf.random.normal([3, 3], mean=0.0, stddev=1.0)
tensor_uniform = tf.random.uniform([5], minval=10, maxval=20)
tensor_shuffled = tf.random.shuffle(tensor_uniform)

# Ép kiểu dữ liệu (Casting)
tensor_cast = tf.cast(rank_0_tensor, dtype=tf.float32)

# ==========================================
# 3. CÁC PHÉP TOÁN TOÁN HỌC & LOGIC
# ==========================================
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

# Toán học cơ bản
c_add = tf.add(a, b)
c_sub = tf.subtract(a, b)
c_mul = tf.multiply(a, b) # Nhân từng phần tử (Element-wise)
c_div = tf.divide(a, b)
mat_mul = tf.matmul(a, b) # Nhân ma trận toán học

# Phép toán logic & Câu lệnh điều kiện (tf.where)
x = tf.constant([1, 2, 3, 4, 5])
y = tf.constant([5, 4, 3, 2, 1])
condition = tf.less(x, y) # Trả về Tensor Boolean [True, True, False, False, False]
result_where = tf.where(condition, x + y, x - y)

# ==========================================
# 4. BIẾN ĐỔI HÌNH DẠNG TENSOR (RESHAPE, CONCAT, TRANSPOSE)
# ==========================================
tensor_to_reshape = tf.constant([[1, 2, 3], [4, 5, 6]])
reshaped_tensor = tf.reshape(tensor_to_reshape, [3, 2])
transposed_tensor = tf.transpose(tensor_to_reshape)

# Nối Tensor (Concatenate và Stack)
t1 = tf.constant([[1, 2], [3, 4]])
t2 = tf.constant([[5, 6], [7, 8]])
concat_axis0 = tf.concat([t1, t2], axis=0) # Nối theo dòng
concat_axis1 = tf.concat([t1, t2], axis=1) # Nối theo cột

# Cắt ảnh / Tensor ngẫu nhiên (Ví dụ giả lập dữ liệu ảnh 3 chiều)
mock_image = tf.random.normal([10, 10, 3])
cropped_image = tf.image.random_crop(mock_image, size=[5, 5, 3])

# ==========================================
# 5. ĐỒ THỊ TÍNH TOÁN (GRAPH MODE)
# ==========================================
@tf.function
def graph_calculation(val1, val2):
    add_val = tf.add(val1, val2)
    return tf.multiply(add_val, 2)

print("\nChạy thử hàm trong Graph Mode:", graph_calculation(tf.constant(5), tf.constant(10)).numpy())

# ==========================================
# 6. XÂY DỰNG MODEL MÔ PHỎNG (LỚP ĐỐI TƯỢNG)
# ==========================================
# Định nghĩa cấu trúc tầng Dense thủ công kế thừa từ tf.Module
class SimpleDense(tf.Module):
    def __init__(self, in_features, out_features, name=None):
        super().__init__(name=name)
        # Khởi tạo ma trận trọng số Weights và Bias ngẫu nhiên bằng tf.Variable
        self.w = tf.Variable(tf.random.normal([in_features, out_features]), name='w')
        self.b = tf.Variable(tf.zeros([out_features]), name='b')

    def __call__(self, x):
        # Công thức tuyến tính: y = xW + b
        y = tf.matmul(x, self.w) + self.b
        return tf.nn.relu(y) # Hàm kích hoạt ReLU

# Khởi tạo mô hình giả định đầu vào 3 tính năng, đầu ra 2 tính năng
my_dense_layer = SimpleDense(in_features=3, out_features=2)

# Chạy thử mô hình với một Tensor đầu vào mẫu
input_data = tf.constant([[2.0, 1.0, 3.0]])
output_data = my_dense_layer(input_data)

print("\n--- KẾT QUẢ KIỂM TRA MÔ HÌNH DENSE LAYER ---")
print("Đầu ra sau khi qua lớp Dense & kích hoạt ReLU:\n", output_data.numpy())
```
